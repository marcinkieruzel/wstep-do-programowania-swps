## Jak pracować z notatnikiem

- Notatnik składa się z **komórek**: tekstowych (jak ta) i z kodem.
- Komórkę z kodem uruchamiasz klawiszami **Shift+Enter** albo przyciskiem ▶ obok komórki. Wynik pojawia się pod spodem.
- Uruchamiaj komórki **po kolei, od góry**. Python pamięta zmienne między komórkami. Jeśli coś się „rozjedzie”, uruchom ponownie jądro (*Restart*) i wykonaj komórki od początku.
- Pod wieloma zadaniami jest **komórka sprawdzająca**: ✅ znaczy „dobrze”, ❌ podpowiada, co poprawić.
- **Błędy są w porządku.** Kilka zadań każe je popełnić celowo — tak najszybciej uczysz się czytać komunikaty Pythona.
- Trudność: ⭐ podstawowe · ⭐⭐ trochę trudniejsze · ⭐⭐⭐ dla chętnych.
- **W Google Colab** najpierw kliknij u góry **Skopiuj na Dysk** (*Copy to Drive*) i pracuj na swojej kopii. Bez tego Twoje odpowiedzi znikną po zamknięciu karty.

Najpierw uruchom komórkę poniżej.

````python
# Komórka przygotowująca — uruchom ją na początku (Shift+Enter).
# Nie musisz rozumieć tego kodu: definiuje funkcję sprawdz(),
# z której korzystają komórki sprawdzające.

def sprawdz(test, podpowiedz="Przeczytaj polecenie jeszcze raz."):
    try:
        wynik = test()
    except NameError as blad:
        print(f"❌ Nie ma jeszcze zmiennej {blad.name}. Uruchom najpierw komórkę z zadaniem.")
        return
    except Exception:
        print(f"❌ Jeszcze nie. {podpowiedz}")
        return
    if wynik:
        print("✅ Dobrze!")
    else:
        print(f"❌ Jeszcze nie. {podpowiedz}")

print("Gotowe — możesz zaczynać.")
````
